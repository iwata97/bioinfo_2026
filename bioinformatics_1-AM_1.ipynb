## 3. データを読み込む

次に、RNA-seq由来の発現量データを読み込みます。

読み込んだ表データを `df_tpm` という名前で保存します。

### TPMとは？

TPMは、RNA-seqで使われる発現量の単位の1つです。

## 10. ヒストグラムを描く

ヒストグラムは、データの分布を見るためのグラフです。  
発現量が小さい遺伝子が多いのか、大きい遺伝子が多いのかを確認できます。

グラフのラベルは、日本語フォントの文字化けを避けるため英語で表示します。

````python
plt.figure(figsize=(6, 4))
plt.hist(x, bins=50)
plt.xlabel("TPM")
plt.ylabel("Number of genes")
plt.title("Distribution of TPM values: batch_1")
plt.show()
````

### 結果の見方

発現量データは、0に近い値が多く、一部に非常に大きな値があります。  
このようなデータは、そのままだとグラフが見にくいことがあります。

そこで、次に **log変換** を行います。

---
## 11. log変換を体験する

発現量のように値のばらつきが大きいデータでは、log変換を行うことがあります。

ここでは、次の式で変換します。

```text
log2(TPM + 1)
```

`+1` しているのは、TPMが0のときにlogを計算できないためです。

````python
log_tpm = np.log2(tpm + 1)

log_tpm.head()
````

log変換後の `batch_1` の分布を見てみましょう。

````python
x_log = log_tpm["batch_1"]

plt.figure(figsize=(6, 4))
plt.hist(x_log, bins=50)
plt.xlabel("log2(TPM + 1)")
plt.ylabel("Number of genes")
plt.title("Distribution after log transformation: batch_1")
plt.show()
````

### 考えてみよう

log変換の前後で、ヒストグラムの見え方はどう変わりましたか。

- 変換前：大きな値に引っ張られやすい
- 変換後：分布全体が見やすくなる

RNA-seq解析では、このようにデータを見やすく・扱いやすくする前処理がよく行われます。

---
## 12. サンプル名を書き換えてグラフを描く

ここからは、少しだけ自分で書き換える練習をします。

下の `sample_name` を変更すると、別のサンプルのヒストグラムを描くことができます。

````python
# ここを書き換えてみましょう。
sample_name = "batch_1"

x_log = log_tpm[sample_name]

plt.figure(figsize=(6, 4))
plt.hist(x_log, bins=50)
plt.xlabel("log2(TPM + 1)")
plt.ylabel("Number of genes")
plt.title("Distribution after log transformation: " + sample_name)
plt.show()
````

## 13. 2つのサンプルを散布図で比べる

次に、2つのサンプルの発現量を散布図で比べます。  
各点は1つの遺伝子を表します。

- 横軸：サンプルAでの発現量
- 縦軸：サンプルBでの発現量

ここでは、解析の詳しい解釈ではなく、**表データから簡単に図を作れること**を体験します。

````python
sample_a = "batch_1"
sample_b = "batch_2"

plt.figure(figsize=(5, 5))
plt.scatter(log_tpm[sample_a], log_tpm[sample_b], s=8, alpha=0.4)
plt.xlabel(sample_a + " log2(TPM + 1)")
plt.ylabel(sample_b + " log2(TPM + 1)")
plt.title(sample_a + " vs " + sample_b)
plt.show()
````

### 書き換えてみよう

上のセルで、`sample_a` と `sample_b` を別のサンプル名に変えてみましょう。

例：

```python
sample_a = "batch_1"
sample_b = "chemostat_1"
```

2つのサンプルを変えるだけで、同じコードから別の図を作ることができます。  
これがNotebookを使った解析の便利なところです。

## 14. データを要約してみる

最後に、各サンプルの平均値を計算してみます。

`mean()` を使うと平均値を計算できます。  
`axis=0` は「列ごとに計算する」という意味です。

````python
sample_means = log_tpm.mean(axis=0)

sample_means
````

平均値を棒グラフにしてみましょう。

````python
plt.figure(figsize=(6, 4))
plt.bar(sample_means.index, sample_means.values)
plt.ylabel("Mean log2(TPM + 1)")
plt.title("Mean expression level of each sample")
plt.xticks(rotation=45)
plt.show()
````
